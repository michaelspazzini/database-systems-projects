# Applicazioni di chat/messagistica

```python
# Preconfigurazione
from pymongo import MongoClient
from datetime import datetime

client = MongoClient("mongodb://localhost:27017/")
db = client["chat_app"]
messages_collection = db["messages_collection"]
```

```python
# Simulazione invio messaggi
messages = [
    {"sender": "Alice", "receiver": "Bob", "message": "Hi Bob, how are you?", "timestamp": datetime.now()},
    {"sender": "Bob", "receiver": "Alice", "message": "I'm good, Thk Alice!", "timestamp": datetime.now()},
    {"sender": "Alice", "receiver": "Bob", "message": "What are you doing later?", "timestamp": datetime.now()},
]

messages_collection.insert_many(messages)
```

```python
def get_conversation(user1, user2):
    conversation = messages_collection.find({
        "$or": [
            {"sender": user1, "receiver": user2},
            {"sender": user2, "receiver": user1},
        ]
    }).sort("timestamp")

    for msg in conversation:
        print(f"{msg['sender']} -> {msg['receiver']}: {msg['message']} ({msg['timestamp']})")

get_conversation("Alice", "Bob")
```
