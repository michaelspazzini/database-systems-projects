# Gestione dei profili e preferenza utente

```python
# Preconfigurazione: connessione
from pymongo import MongoClient
import random
import time
from datetime import datetime

client = MongoClient("mongodb://localhost:27017/")
db = client["users"]
profiles_collection = db["profiles_collection"]
```

```python
db.drop_collection("profiles_collection")
```

```python
# Creazione profilo utente con preferenze
user_profile ={
    "username" : "spamic",
    "email": "michaelspazzini@gmail.com",
    "preference":{
        "theme" : "dark",
        "notifications" : True,
        "languages":[
            "English",
            "Italian"
        ]
    },
    "order_history": [
        {"order_id": 1, "product": "Laptop", "amount": 1200, "date":"2024-09-01"},
        {"order_id": 2, "product": "Smartphone", "amount": 800, "date":"2024-10-01"}
    ]
}

# Inserimento profilo su db
profiles_collection.insert_one(user_profile)
print("Profilo utente inserito")
```

```python
# Funzione per aggiornare le preferenze utente
def update_user_preferences(username, new_preferences):
    profiles_collection.update_one(
        {"username": username},
        {"$set": {
            "preferences": new_preferences
        }}
    )
    print(f"Preferenze aggiornate per utente: {username}")
```

```python
# Funzione per recuperare la cronologia deli ordini dell'utente
def get_user_order_history(username):
    user = profiles_collection.find_one({"username": username})
    if user:
        for order in user["order_history"]:
            print(f"Ordine {order['order_id']}:  {order['product']} -  {order['amount']} USD,  {order['date']}")

    else:
        print("Utente non trovato")
```

```python
# Aggiorniamo le preferenze di spamic
new_preferences = {
    "theme" : "light",
    "notification": False,
    "languages": ["English", "Italian"]
}
update_user_preferences("spamic", new_preferences)
```

```python
# Visualizzazione cronologia ordini spamic
get_user_order_history("spamic")
```
