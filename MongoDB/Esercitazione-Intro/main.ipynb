# Studio di MongoDB

```python
# Preconfiguration: connection
import pymongo
from pymongo import MongoClient
import datetime

myclient = MongoClient("mongodb://localhost:27017/")
mydb = myclient["post"]
mycol = mydb["post"]
```

```python
# Insert one
myPost = {
    "title": "Welcome to the malboro Part.2",
    "slug": "welcome-to-malborto-part-2",
    "body": "smoke smoke smoke...",
    "published": True,
    "created": datetime.datetime.now().strftime("%a %b %d %Y %H:%M:%S"),
    "updated": datetime.datetime.now().strftime("%a %b %d %Y %H:%M:%S"),
    "comments": [
        {
            "author": "Michael",
            "email": "michael.spazzini@gmail.com",
            "body": "Don't smoke!",
            "created": datetime.datetime.now().strftime("%a %b %d %Y %H:%M:%S"),
        }
    ],
    "tags": ["databases", "MongoDB", "awesome"],
    "rate": 5
}

insertedDoc = mycol.insert_one(myPost)
print(f"Post inserito: {insertedDoc.inserted_id}")
```

```python
# Insert Many
newPosts = [
    {
        "title": "Message Number 4",
        "slug": "msg-4",
        "body": "Hey, come va?",
        "published": False,
        "created": datetime.datetime.now().strftime("%a %b %d %Y %H:%M:%S"),
        "updated": datetime.datetime.now().strftime("%a %b %d %Y %H:%M:%S"),
        "tags": ["databases", "MongoDB", "awesome"],
        "rate": 1
    },
    {
        "title": "Message Number 5",
        "slug": "msg-5",
        "body": "Bene, e te?",
        "published": True,
        "created": datetime.datetime.now().strftime("%a %b %d %Y %H:%M:%S"),
        "updated": datetime.datetime.now().strftime("%a %b %d %Y %H:%M:%S"),
        "tags": ["databases", "MongoDB", "awesome"],
        "rate": 3
    }
]

results = mycol.insert_many(newPosts)
print(f"Inserted post: {results.inserted_ids}")
```

# Modify, insert, remove

```python
# update documents
# db.collection.updateOne(<filter>,[<update>])
# db.collection.updateMany(<filter>,[<update>])
# db.collection.replaceOne(<filter>,[<update>])

# Aggiornamento multiplo
mydb.post.update_many(
    {"published": True},
    {"$set": {"published": False}}
)
```

```python
# Add new field
mydb.post.update_many(
    {},
    [{
        "$addFields": {
        "newRate" : { 
            "$multiply": [
                "$rate", 10
                ]
            }
        }
    }]
)
```

```python
# Multple remove
mydb.post.update_many(
    {},
    [
        {
            "$unset": ["newRate"]
        }
    ]
)
```

# Remove documents

```python
# Remove only the first document that satisfy filter condition 
mydb.post.delete_one({"published":True})
```

```python
# Remove all documents thaht satisfy filter condition
mydb.post.delete_many({"slug":"msg-3"})
```

# Data validation

```python
mydb.create_collection("validatedPost", validator={
    "$jsonSchema":{
        "bsonType": "object",
        "title": "Post object Validation",
        "required": ["title", "body", "published", "rate"],
        "properties":{
            "name" : {
                "bsonType" : "string",
                "description": "Optional description"
            },
            "body" : {
                "bsonType" : "string",
                "description": "Optional description"
            },
            "published" : {
                "bsonType" : "string",
                "description": "Optional description"
            },
            "rate" : {
                "bsonType" : "int",
                "minimum": 0,
                "maximum": 5,
                "description": "Optional description"
            },
        }
    }
})
```
